```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.svm import SVC
```

```python
X, y = make_blobs(n_samples = 100, centers = 2, random_state = 42, cluster_std = 1.5)
```

```python
svm_linear = SVC(kernel = 'linear', C = 1.0)
svm_linear.fit(X, y)
```

```python
svm_rbf = SVC(kernel='rbf', C = 1.0, gamma = 'scale')
svm_rbf.fit(X, y)
```

```python
x_min, x_max = X[:, 0].min() - 1, X[:,0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:,1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))
```

```python
fig, axes = plt.subplots(1, 2, figsize = (14, 6))
models = [('SVM Linear Kernel', svm_linear), ('SVM RBF Kernel', svm_rbf)]
for ax, (name, model) in zip(axes, models):
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha = 0.3, cmap = 'RdBu_r')
    ax.contour(xx, yy, Z, levels = [0.5], colors = 'black', linewidths = 2, linestyles = '--')
    ax.scatter(model.support_vectors_[:, 0], model.support_vectors_[:, 1], s= 120, facecolors = 'none', edgecolors= 'k', linewidths = 1.5, label = 'Support Vectors')

    ax.scatter(X[y == 0][:, 0], X[y == 0][:, 1], color = 'blue', label = 'Class 0', edgecolors = 'k')
    ax.scatter(X[y == 1][:, 0], X[y == 1][:, 1], color = 'red', label = 'Class 1', edgecolors = 'k')

    ax.set_title(f'Decision Boundary: {name}')
    ax.set_xlabel('Feature 1 (x1)')
    ax.set_ylabel('Feature 2 (x2)')
    ax.legend(loc = 'lower left')
    ax.grid(True, linestyle)
    
```

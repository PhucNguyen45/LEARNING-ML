```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_circles, make_moons
from sklearn.neighbors import KNeighborsClassifier
```

```python
# 1. Sinh 3 dạng dữ liệu (2 lớp) khác nhau
np.random.seed(42)

# Dạng 1: Phân tách tuyến tính (Linearly Separable)
X_linear, y_linear = make_blobs(n_samples=200, centers=2, random_state=42, cluster_std=1.2)

# Dạng 2: Phân tách phi tuyến dạng Đa thức / Trăng khuyết (Moons)
X_poly, y_poly = make_moons(n_samples=200, noise=0.2, random_state=42)

# Dạng 3: Phân tách phi tuyến dạng Hình tròn đồng tâm (Circles)
X_circle, y_circle = make_circles(n_samples=200, noise=0.1, factor=0.4, random_state=42)

datasets = [
    ("1. Tuyến tính (Linearly Separable)", X_linear, y_linear),
    ("2. Phi tuyến dạng Đa thức (Moons)", X_poly, y_poly),
    ("3. Phi tuyến dạng Hình tròn (Circles)", X_circle, y_circle)
]
```

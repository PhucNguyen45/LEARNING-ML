EX2: Given the dataset of human’s weight and height, write a Python program to estimate the weight given the height, e.g.155 and 160 (cm)?

dataset:
Height(cm): 147 150 153 155 158 160 163 165 168 170 173 175 178 180 183
Weight(kg): 49  50  51  52  54  56  58  59  60  62  63  64  66  67  68

```python
# import important libraries 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# height (cm)
X_org = np.array([[147, 150, 153, 158, 163, 165, 168, 170, 173, 175, 178, 180, 183]]).T
# weight (kg)
y = np.array([[ 49, 50, 51,  54, 58, 59, 60, 62, 63, 64, 66, 67, 68]]).T
```

```python
# Visualize data
plt.scatter(X_org, y)
plt.xlabel('x = Height (cm)')
plt.ylabel('y = Weight (kg)')
plt.grid()
plt.show()
```

```python
one = np.ones((13,1))
one.shape
```

```python
X = np.concatenate((one, X_org), axis=1)
X
```

```python
theta = np.linalg.inv(X.T @ X) @ X.T @ y
theta
```

```python
y_pred = X @ theta
y_pred
```

```python
# Visualize data and prediction
plt.scatter(X_org,y)
plt.plot(X_org, y_pred)
plt.xlabel('x = Height (cm)')
plt.ylabel('y = Weight (kg)')
plt.grid()
plt.show()
```

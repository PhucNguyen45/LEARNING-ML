```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {"age":           ["youth","youth","middle_aged","senior","senior","senior",
                          "middle_aged","youth","youth","senior","youth",
                           "middle_aged","middle_aged","senior"],
        "income":        ["high","high","high","medium","low","low","low","medium",
                           "low","medium","medium","medium","high","medium"],
        "student":       ["no","no","no","no","yes","yes","yes","no","yes","yes",
                           "yes","no","yes","no"],
        "credit_rating": ["fair","excellent","fair","fair","fair","excellent","excellent",
                           "fair","fair","fair","excellent","excellent","fair","excellent"],
        "buys_computer": ["no","no","yes","yes","yes","no","yes","no","yes","yes",
                           "yes","yes","yes","no"],
}
```

```python
df = pd.DataFrame(data)
df
```

```python
classes = list(df[df.columns[-1]].unique())
n_classes = len(classes)
n_classes, classes
```

```python
#1 compute prior probabilities
p_prior = {c: sum(df[df.columns[-1]] == c) / len(df) for c in classes}
p_prior
```

```python
#2 compute likelihoods
features = list(df.columns[:-1])
features
```

```python
p_likelihood = {}

for c in classes:
    dfc = df[df[df.columns[-1]] == c]

    for feat in features:
        feat_vals = list(dfc[feat].unique())

        for val in feat_vals:
            p_likelihood[f'{feat} = {val} | {c}'] = sum(dfc[feat] == val) / len(dfc)

p_likelihood
```

```python
# Inference
x = {'age = youth','income = medium','student = yes','credit_rating = fair'}
p = {c: 1 for c in classes}
# c = classes[0]
for c in classes:
    # Compute likelihood of x given c
    for xi in x:
        p[c]*= p_likelihood[f'{xi} | {c}']
    # Compute product of likelihood with prior prob
    p[c] *= p_prior[c]
# Predict the label with the maximum product
y_pred = max(p, key = p.get)
p, y_pred
```
